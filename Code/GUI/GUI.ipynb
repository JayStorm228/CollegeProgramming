# Модуль 1. Урок №1: Анатомия GUI и магия Событийного цикла
Добро пожаловать на первый этап становления GUI Архитектора. Сегодня мы заложим фундамент: поймем, чем программа с интерфейсом отличается от обычного скрипта, и создадим наше первое «правильное» окно.

## 1. Теоретический блок (Глоссарий)
Для начала определимся с терминами, которые станут твоим ежедневным языком:  
- `GUI` (Graphical User Interface) — графический интерфейс пользователя. Система визуальных компонентов (окон, кнопок, меню), позволяющая человеку взаимодействовать с кодом не через текст, а через образы.
- `Widget` (Виджет) — «штуковина» или элементарный объект интерфейса. Кнопка, текстовое поле, чекбокс — всё это виджеты.
- `Even Loop` (Событийный цикл) — бесконечный цикл, который «слушает» действия пользователя (клики, нажатия клавиш) и сообщает программе, что пора реагировать.
- `Callback` (Обратный вызов) — функция, которую мы привязываем к событию. Она «спит» до тех пор, пока пользователь не нажмет на кнопку

---

## 2. Архитектура и Логика: «Под капотом»
В обычном Python-скрипте код выполняется сверху вниз и завершается. В GUI всё иначе.  
Почему обычный `while True` — это яд для интерфейса? Если ты напишешь свой цикл для обновления окна, программа «зависнет». Она будет так занята твоим циклом, что не успеет обработать системные сигналы: «отрисовать курсор», «изменить цвет кнопки при наведении» или «закрыть окно».  
**Решение**: Метод `mainloop()`. Это высокооптимизированный событийный цикл. Он работает так быстро, что создается иллюзия мгновенной реакции

---

## 3. UI/UX Разбор: Психология восприятия
- `UI` (User Interface) — это «лицо». Красивые скругления и цвета.
- `UX` (User Experience) — это «ощущение». Насколько легко пользователю достичь цели.
  - Базовое правило визуального веса: Глаз пользователя в первую очередь падает на самый контрастный и крупный элемент.
  - Совет: Центр пустого окна — это «мертвая зона». Внимание всегда начинается с левого верхнего угла

# Урок №2: Иерархия объектов и управление «детьми»
На прошлом уроке мы создали фундамент — главное окно. Сегодня мы научимся наполнять его элементами. В GUI элементы не существуют сами по себе: они всегда находятся в строгих семейных отношениях.

---

## 1. Теоретический блок (Глоссарий)
- Родитель (Parent/Master) — объект-контейнер (например, окно или фрейм), внутри которого «живут» другие элементы.
- Ребенок (Child) — виджет, который привязан к родителю. Если уничтожить родителя, все его «дети» исчезнут автоматически.
- Дерево виджетов (Widget Tree) — иерархическая структура данных, где на вершине стоит один «корень» (Root), а от него ветвятся остальные компоненты.
- Фрейм (Frame) — невидимый прямоугольный контейнер, используемый исключительно для логической и визуальной группировки других виджетов.

---

## 2. Архитектура и Логика: «Дерево элементов»
В разработке интерфейсов властвует иерархия. Каждый раз, когда ты создаешь виджет, ты обязан указать, кто его «хозяин».
Зачем это нужно?
- Управление ресурсами: Закрыл окно — Python сам очистил память от всех кнопок внутри него.
- Геометрия: Позиция ребенка всегда считается относительно левого верхнего угла его родителя, а не всего экрана.
**Визуальная схема иерархии (Widget Tree)**:
Root Window (Главное окно)  
```
 └── Main Frame (Главный контейнер)  
      ├── Header Label (Текст заголовка)  
      └── Content Frame (Зона контента)  
           └── Action Button (Кнопка действия)  
```
---

Каждый уровень вложенности делает интерфейс более структурированным и легким для рефакторинга.
## 3. UI/UX Разбор: Визуальная иерархия
- Золотое правило дизайна: Визуальная иерархия должна отражать логическую структуру программы.
- Контраст: Родительские контейнеры (заголовки групп) должны быть менее яркими, чем кнопки действий (дети), чтобы не отвлекать пользователя от цели.
- Группировка: Похожие элементы должны находиться в одном родительском контейнере (Frame). Это дает пользователю психологическое ощущение порядка

## 4. Код (Clean OOP)
Мы создадим окно, где внутри главного окна живет «Фрейм», а внутри него — «Надпись».


````python
import tkinter as tk
from tkinter import ttk

class HierarchyApp(tk.Tk):
    def __init__(self):
        super().__init__()
        self.title("GUI Architect: Урок 2")
        self.geometry("400x200")

        # Запуск создания иерархии
        self._setup_ui()

    def _setup_ui(self):
        # 1. Создаем родительский контейнер (Фрейм)
        # Указываем self как родителя
        self.container = tk.Frame(self, bg="#f0f0f0", bd=2, relief="groove")
        self.container.pack(padx=20, pady=20, fill="both", expand=True)

        # 2. Создаем ребенка для Фрейма (Надпись)
        # Указываем self.container как родителя!
        self.label = tk.Label(self.container, text="Я ребенок внутри Фрейма", bg="#f0f0f0")
        self.label.pack(pady=10)

        # 3. Создаем еще одного ребенка для Фрейма (Кнопка)
        self.button = tk.Button(self.container, text="Удалить контейнер", command=self._on_destroy_click)
        self.button.pack()

    def _on_destroy_click(self):
        """Метод-обработчик события (Callback)."""
        print("Удаляем родителя — дети исчезнут вместе с ним.")
        self.container.destroy() # Уничтожаем фрейм

if __name__ == "__main__":
    app = HierarchyApp()
    app.mainloop()
````

## 5. Построчный анализ
- `tk.Frame(self, ...)`: Мы создаем контейнер и передаем self (наш класс окна) в качестве родителя.
- `tk.Label(self.container, ...)`: Здесь магия вложенности. Мы указываем, что надпись принадлежит не окну напрямую, а нашему фрейму.
- `self.container.pack(...)`: Мы приказываем родителю разместить себя в окне. Пока родитель не «размещен» (gridded/packed), его дети не будут видны.
- `self.container.destroy()`: Демонстрация власти иерархии. Вызов этого метода каскадно уничтожает и кнопку, и надпись, потому что они — его «дети»

# Урок №3: Геометрия интерфейса и концепция «Коробки» (Метод pack)
Сегодня мы разберем один из самых интуитивных, но часто недооцененных способов размещения элементов — метод pack(). Он превращает ваше окно в набор вложенных коробок, которыми нужно уметь управлять

---

## 1. Теоретический блок (Глоссарий)
- `Geometry` Manager (Менеджер геометрии) — механизм, отвечающий за расчет размеров и позиций виджетов в окне.
- `Packer` (Упаковщик) — менеджер геометрии, который размещает виджеты, «прижимая» их к одной из сторон родительского контейнера.
- `Box Model` (Коробочная модель) — концепция, согласно которой каждый виджет рассматривается как прямоугольный блок, занимающий определенное пространство.
- `Padding` (Отступы) — пустое пространство вокруг виджета:
  - `External` (`padx`/`pady`) — внешние отступы (от границы виджета до соседей).
  - `Internal` (`ipadx`/`ipady`) — внутренние отступы (от границы виджета до его содержимого).

---

## 2. Архитектура и Логика: «Принцип очереди»
Под капотом упаковщик работает по принципу выделения свободного пространства.  
Когда вы вызываете `pack()`, менеджер:
- Берет оставшуюся свободную область родителя.
- Отрезает от нее кусок вдоль указанной стороны (`side`).
- Помещает туда виджет и больше не позволяет другим виджетам занимать это место.
Параметры управления пространством:
- `side`: К какому краю прижать (`TOP`, `BOTTOM`, `LEFT`, `RIGHT`).
- `fill`: Должен ли виджет растягиваться, чтобы заполнить выделенную ему «коробку» (`X`, `Y` или `BOTH`).
- `expand`: Если `True`, виджет будет забирать себе всё лишнее свободное место, которое осталось в родителе.

--------------------------------------------------------------------------------
## 3. UI/UX Разбор: Воздух в интерфейсе
- Белое пространство (White Space) — это не пустота, а инструмент управления вниманием.  
- **Правило близости**: Элементы, расположенные близко друг к другу, воспринимаются пользователем как связанные по смыслу.  
- Зачем нужны Padding/Margin? Если кнопки «прилипают» к краям окна или друг к другу, интерфейс кажется перегруженным и вызывает «клаустрофобию» у пользователя. Оптимальные отступы снижают когнитивную нагрузку.

---

## 4. Код (Clean OOP)
Создадим панель управления с «правильным» распределением пространства.

````python
import tkinter as tk

class LayoutApp(tk.Tk):
    def __init__(self):
        super().__init__()
        self.title("GUI Architect: Урок 3")
        self.geometry("400x300")
        self._setup_ui()

    def _setup_ui(self):
        # Верхний заголовок - прижат к верху, растянут по горизонтали
        self.header = tk.Label(self, text="Панель инструментов", bg="#000333", fg="white")
        self.header.pack(side="top", fill="x", ipady=10) # ipady дает внутреннюю высоту

        # Боковая панель
        self.sidebar = tk.Frame(self, bg="#000eee", width=100)
        self.sidebar.pack(side="left", fill="y")

        # Кнопки на боковой панели с внешними отступами
        tk.Button(self.sidebar, text="Профиль").pack(pady=10, padx=10, fill="x")
        tk.Button(self.sidebar, text="Настройки").pack(pady=10, padx=10, fill="x")

        # Центральная область, которая забирает ВСЁ оставшееся место
        self.content = tk.Label(self, text="Основной контент", bg="white")
        self.content.pack(side="right", expand=True, fill="both")

if __name__ == "__main__":
    app = LayoutApp()
    app.mainloop()
````

## 5. Построчный анализ
- `ipady=10`: Мы увеличили «высоту» заголовка за счет внутренних отступов, не меняя шрифт.
- `side="left", fill="y"`: Фрейм `sidebar` занял всю левую часть окна сверху донизу.
- `pady=10, padx=10`: Мы создали «безопасную зону» вокруг кнопок, чтобы они не касались краев фрейма.
- `expand=True, fill="both"`: Магическая связка. `expand` говорит: «Забери всё лишнее место в центре», а`fill` — «Растянись и заполни его визуально»

## 6. Практикум
- Базовое задание: Создай окно с тремя горизонтальными полосами (Фреймами) разного цвета. Каждая должна занимать 1/3 высоты окна при любом его расширении (используй `expand=True` и `fill='both`').
- Challenge (UX): Размести внутри нижнего фрейма две кнопки: одну слева, другую справа. Сделай так, чтобы при расширении окна кнопки оставались у своих краев. Подсказка: используй разные значения `side` для кнопок внутри одного родителя.

````python
import tkinter as tk
from typing import Any

def loop_array_front(array: list[Any], start:int):
    index = start
    if index+1 >= len(array):
        index = 0
    index += 1
    return array[index]

def loop_array_back(array: list[Any], start:int):
    index = start
    if index-1 < 0:
        index = len(array)
    index -=1
    return array[index]

class LayoutApp(tk.Tk):
    def __init__(self, width: int, height:int):
        super().__init__()
        self.title("GUI Architect: Урок 3")
        self.width, self.height = width, height
        self.geometry(f"{width}x{height}")
        self.colors = [
            "#000000",
            "#FFFFFF",
            "#666666",
            "#E74C3C",
            "#E67E22",
            "#F1C40F",
            "#2ECC71",
            "#3498DB",
            "#2980B9"
        ]
        self._setup_ui()
    def _setup_ui(self):
        self.header = tk.Label(
            self,
            text="Программа",
            bg="#000333",
            fg="white"
        )
        self.header.pack(
            side="top",
            fill="x",
            ipady=5
        )
        self._setup_TopFrame()
        self._setup_MidFrame()
        self._setup_BotFrame()

    def _setup_TopFrame(self):
        self.TopFrame = tk.Frame(
            self,
            height=round(0.33*self.height),
            bg=self.colors[0]
        )
        self.TopFrame.pack(
            side="top",
            fill="x",
            expand=True,
            pady=1
        )
        self.TopFrame_Button1 = tk.Button(
            self.TopFrame,
            text="ChangeColor \nforward",
            bg = "#F6F6F6",
            fg = "#000000",
            command=self._ChangeColor_TopFrame_front
        )
        self.TopFrame_Button1.pack(
            side="left",
            ipady=1,
            pady=5
        )
        self.TopFrame_Button2 = tk.Button(
            self.TopFrame,
            text="ChangeColor \nbackward",
            bg = "#F6F6F6",
            fg = "#000000",
            command=self._ChangeColor_TopFrame_back
        )
        self.TopFrame_Button2.pack(
            side="right",
            ipady=1,
            pady=5
        )

    def _ChangeColor_TopFrame_front(self):
        CurrentColorID = self.colors.index(self.TopFrame["bg"])
        self.TopFrame["bg"] = loop_array_front(self.colors, CurrentColorID)

    def _ChangeColor_TopFrame_back(self):
        CurrentColorID = self.colors.index(self.TopFrame["bg"])
        self.TopFrame["bg"] = loop_array_back(self.colors, CurrentColorID)

    def _setup_MidFrame(self):
        self.MidFrame = tk.Frame(
            self,
            height=round(0.33*self.height),
            bg="#eeeeee",
        )
        self.MidFrame.pack(
            side="top",
            fill="x",
            expand=True,
            pady=1
        )
    def _setup_BotFrame(self):
        self.BotFrame = tk.Frame(
            self,
            height=round(0.33*self.height),
            bg="#000000",
        )
        self.BotFrame.pack(
            side="top",
            fill="x",
            expand=True,
            pady=1
        )
if __name__ == "__main__":
    app = LayoutApp(
        width=500,
        height=500
    )
    app.mainloop()
````

# Модуль 2. Урок №4: Текстовые блоки и кнопки (`Label`, `Button`)
Мы переходим к изучению «лица» любого интерфейса. Если Модуль 1 был посвящен каркасу, то Модуль 2 научит нас работать с конкретными органами чувств программы: как она говорит с пользователем и как она его слушает

---

## 1. Теоретический блок (Глоссарий)
- `Label` (Надпись) — статичный виджет, предназначенный для отображения текста или изображений, которые пользователь не может редактировать напрямую.
- `Button` (Кнопка) — активный виджет, который «проседает» при нажатии и выполняет привязанное к нему действие.
- `Affordance` (Возможность) — свойство объекта, подсказывающее, как с ним взаимодействовать. Кнопка «просит», чтобы на неё нажали, благодаря визуальному объему.
- `Command` (Команда) — параметр виджета (обычно кнопки), в который передается ссылка на функцию или метод, вызываемый при активации элемента

---

## 2. Архитектура и Логика: «Система стимул-реакция»
Под капотом взаимодействие с кнопкой работает через механизм регистрации обратного вызова (`callback`):
1. При создании объекта `tk.Button` мы передаем в аргумент `command` имя нашего метода: `self._my_method`.
2. Важно: Мы передаем ссылку на метод (без скобок), а не вызываем его сразу.
3. Когда пользователь кликает на кнопку, `Event Loop` (событийный цикл) ловит этот сигнал и принудительно вызывает привязанный метод.  

**Дерево атрибутов**: Каждый виджет — это объект с десятками свойств (цвета, шрифты, отступы).  
Мы можем менять их как при создании, так и на лету через метод `.configure()`

**Шрифты и иерархия**:

- **Психология кнопки**: Текст на кнопке должен начинаться с глагола действия (`«Сохранить»`, `«Удалить»`, `«Начать»`), чтобы пользователь сразу понимал результат клика.

## 4. Код (Clean OOP)
Создадим интерактивный блок, где кнопка меняет состояние информационной надписи.

````python
import tkinter as tk

class InteractionApp(tk.Tk):
    def __init__(self):
        super().__init__()
        self.title("GUI Architect: Урок 4")
        self.geometry("300x200")
        self._setup_ui()

    def _setup_ui(self):
        # Создаем надпись с настройкой шрифта (Label)
        # font=('Название', размер, 'стиль')
        self.info_label = tk.Label(
            self,
            text="Ожидание команды...",
            font=("Verdana", 12),
            pady=20
        )
        self.info_label.pack()

        # Создаем интерактивную кнопку (Button)
        self.action_btn = tk.Button(
            self,
            text="Выполнить действие",
            font=("Arial", 10, "bold"),
            bg="#2ecc71", # Зеленый цвет
            fg="white",   # Белый текст
            command=self._handle_click # Ссылка на метод
        )
        self.action_btn.pack(ipadx=10, ipady=5)

    def _handle_click(self):
        """Метод-обработчик события (Reaction)."""
        self.info_label.configure(
            text="Команда выполнена!",
            fg="green"
        )
        self.action_btn.configure(
            state="disabled", # Выключаем кнопку после нажатия
            text="Завершено"
        )

if __name__ == "__main__":
    app = InteractionApp()
    app.mainloop()
````

## 5. Построчный анализ
- `font=("Verdana", 12)`: Задает семейство и размер. В `Tkinter` это кортеж, который легко читать и править.  
- `bg="#2ecc71", fg="white"`: Настройка цветовой схемы через `HEX`-коды для современного вида  
- `command=self._handle_click`: Привязка логики. Здесь нет скобок в конце, иначе код выполнится сам при запуске программы  
- `self.info_label.configure(...)`: Главный инструмент динамики. Мы меняем свойства уже существующего объекта в памяти  
- `state="disabled"`: Переводит виджет в неактивное состояние. Визуально он становится серым и перестает реагировать на клики

# Урок №5: Ввод данных (Entry) и основы валидации
Теперь мы научим программу «слушать» пользователя через текст. Поле ввода — это не просто пустая строка, это входные ворота для данных, которые нужно уметь защищать и проверять.

## 1. Теоретический блок (Глоссарий)
- `Entry` (Поле ввода) — виджет, предназначенный для ввода и редактирования одной строки текста.
- `Validation` (Валидация) — процесс проверки вводимых данных на соответствие заданным правилам (например, только цифры).
- `StringVar` — специальная управляющая переменная `Tkinter`, которая автоматически синхронизируется с текстом внутри виджета (`Data Binding`).
- `Substitution Code` (Код подстановки) — специальные символы (например, `%P`, `%S`), которые передают в функцию валидации информацию о том, что именно и куда пользователь пытается вписать.

---

## 2. Архитектура и Логика: «Система контроля границ»
В разработке интерфейсов есть два пути получения данных из поля ввода:
- Активный: Вызов метода `.get()` в нужный нам момент (например, при нажатии кнопки).
- Реактивный: Использование `StringVar`. Мы привязываем переменную к полю через опцию `textvariable`, и любое изменение в поле мгновенно меняет значение переменной.

Механизм валидации работает как фильтр:
1. Сначала мы создаем функцию, которая возвращает `True` (разрешить ввод) или `False` (запретить).
2. Затем мы «регистрируем» эту функцию в системе через метод `.register()`, превращая её в понятный для ядра `Tkinter` (`Tcl`) объект.
3. Привязываем регистрацию к виджету через `validatecommand`.

---

## 3. UI/UX Разбор: Психология заполнения форм
Как не напугать пользователя?
1. **Фокус внимания**: 
  - Используй метод `.focus_set()`, чтобы курсор сразу стоял в первом поле. 
  - Пользователю неприятно делать лишний клик, чтобы просто начать писать.
2. **Подсказки (`Placeholders`)**: 
  - Пустое белое поле выглядит пугающе. 
  - Всегда подписывай поля через `Label `или устанавливай начальное значение (через `.insert()`), которое исчезает при клике.
3. **Тихая валидация**: 
  - Лучше запретить ввод буквы в поле «Возраст» сразу (%S), чем позволить пользователю написать всё, нажать «ОК» и выдать ему красную ошибку.

---

## 4. Код (Clean OOP)
Создадим форму, которая принимает только цифровой код

````python
import tkinter as tk

class SecureEntryApp(tk.Tk):
    def __init__(self):
        super().__init__()
        self.title("GUI Architect: Урок 5")
        self.geometry("300x150")
        self._setup_ui()

    def _setup_ui(self):
        # Описание поля
        tk.Label(self, text="Введите PIN-код (только цифры):").pack(pady=(10, 0))

        # Регистрация валидации (Важный архитектурный шаг)
        # Мы говорим: "Используй мой метод _validate_numeric и передай в него %P и %S"
        vcmd = (self.register(self._validate_numeric), '%P', '%S')

        # Создание поля ввода с валидацией
        self.code_entry = tk.Entry(
            self,
            validate='key',           # Проверка при каждом нажатии клавиши
            validatecommand=vcmd,     # Наша зарегистрированная команда
            font=("Courier", 14),
            justify="center"          # Текст будет по центру
        )
        self.code_entry.pack(pady=10, padx=20, fill="x")
        self.code_entry.focus_set()   # Автофокус при старте

    def _validate_numeric(self, current_val:str, char_to_insert:str):
        """Метод проверки: возвращает True, если ввод разрешен."""
        # Разрешаем ввод только если вставляемый символ — цифра
        # и общая длина не превышает 4 знака
        if char_to_insert.isdigit() and len(current_val) <= 4:
            return True
        return False

if __name__ == "__main__":
    app = SecureEntryApp()
    app.mainloop()
````

## 5. Построчный анализ
- `self.register(self._validate_numeric)`: Оборачивает Python-метод в `Tcl`-функцию, чтобы ядро `GUI` могло вызывать её **мгновенно** при каждом нажатии клавиши.
- `validate='key'`: Указывает, что проверка должна срабатывать именно при изменении содержимого (нажатии клавиш), а не при потере фокуса.
- `'%P'` и `'%S'`:  Специальные коды.  
  1. `%P` — это то, как будет выглядеть текст, если мы разрешим ввод.  
  2. `%S` — это конкретный символ, который пользователь только что нажал. 
- `justify="center"`: Визуальная настройка. Для коротких кодов (`PIN`, `ID`) центрирование текста повышает удобство считывания.
- `self.code_entry.focus_set()`: Передача клавиатурного фокуса виджету. Теперь пользователь может сразу начать печатать

````python
import string
import tkinter as tk

colors = {
    "black": "#000000",
    "white": "#FFFFFF",
    "gray": "#666666",
    "red": "#E74C3C",
    "orange": "#E67E22",
    "yellow": "#F1C40F",
    "green": "#2ECC71",
    "white-blue": "#3498DB",
    "blue": "#2980B9",
}
alphabet = string.ascii_letters

class InteractionApp(tk.Tk):
    def __init__(self):
        super().__init__()
        self.title("GUI Architect: Урок 4")
        self.geometry("800x600")
        self._setup_ui()

    # Setup
    def _setup_ui(self):
        self._setup_Heading()
        self._setup_Name_Entry()
        self._setup_Age_Entry()
        self._setup_PIN_Entry()
        self._setup_CheckPIN_Button()
        self._setup_ConditionsCheckBox()
        self._setup_Registration_Button()

    def _setup_Heading(self):
        self.Heading = tk.Label(
            self, text="User Registration", font=("Verdana", 12), pady=20
        )
        self.Heading.pack()
        self.RegistrationFrame = tk.Frame(self, bg="#666666")
        self.RegistrationFrame.pack(fill="both", expand=True)


    # Entries
    def _setup_Name_Entry(self):
        tk.Label(
            master=self.RegistrationFrame,
            text='Username: '
        ).pack(pady=5)
        vcmd = self.register(self._validate_alphabetic), '%P', '%S'

        self.UserName_Entry = tk.Entry(
            master=self.RegistrationFrame,
            validate="key",
            validatecommand=vcmd,     # Наша зарегистрированная команда
            font=("Courier", 14),
            justify="left"
        )
        self.UserName_Entry.pack(pady=5)

    def _setup_Age_Entry(self):
        tk.Label(
            master=self.RegistrationFrame,
            text='Age: '
        ).pack(pady=5)
        vcmd = self.register(self._validate_Age), '%P', '%S'

        self.Age_Entry = tk.Entry(
            master=self.RegistrationFrame,
            validate="key",
            validatecommand=vcmd,
            font=("Courier", 14),
            justify="left",
        )
        self.Age_Entry.pack(pady=5)

    def _setup_PIN_Entry(self):
        tk.Label(
            master=self.RegistrationFrame,
            text='Pin: '
        ).pack(pady=5)

        vcmd = (self.register(self._validate_PIN), '%P', '%S')
        self.PIN_Entry = tk.Entry(
            master=self.RegistrationFrame,
            validate="key",
            validatecommand=vcmd,
            font=("Courier", 14),
            justify="left",
            show="*"
        )
        self.PIN_Entry.pack(pady=5)


    # Buttons
    def _setup_ConditionsCheckBox(self):

        self.ConditionsCheckBox = tk.Button(
            self.RegistrationFrame,
            bg=colors["white"],
            fg=colors["black"],
            text="I agree with terms",
            command=self._handle_CheckBox_Button,
        )
        self.ConditionsCheckBox.pack(pady=5)

    def _setup_Registration_Button(self):
        self.RegistrationButton = tk.Button(
            self.RegistrationFrame,
            bg=colors["white"],
            fg=colors["black"],
            text="Register User",
            state="disabled",
            command=self._handle_Registration_Button,
        )
        self.RegistrationButton.pack(pady=5)

    def _setup_CheckPIN_Button(self):
        self.CheckPIN_Button = tk.Button(
            master=self.RegistrationFrame,
            bg=colors["white"],
            fg=colors["black"],
            text="Check entered PIN",
            command=self._handle_CheckPIN_Button
        )
        self.CheckPIN_Button.pack(
            pady=5
        )
    # Handlers
    def _handle_CheckBox_Button(self):

        self.ConditionsCheckBox.configure(
            state="disabled",
            text="Agreed",
            fg=colors["green"],
        )
        self.RegistrationButton.configure(state="active")

    def _handle_Registration_Button(self):
        self.RegistrationButton.configure(
            state="disabled", text="Successfully registered!", fg=colors["green"]
        )
        self.UserName = self.UserName_Entry.get()
        self.Age = self.Age_Entry.get()
        self.PIN = self.PIN_Entry.get()

        self.UserName_Entry.configure(state='disabled')
        self.Age_Entry.configure(state='disabled')
        self.PIN_Entry.configure(state='disabled')
        print(f'''
User:
Name = {self.UserName if self.UserName else "None"}
Age = {self.Age if self.Age else "None"}
PIN = {self.PIN if self.PIN else "None"}
            '''.strip())
    def _handle_CheckPIN_Button(self):
        print(f"PIN: {self.PIN_Entry.get()}")

    # Validators
    def _validate_Age(self, current_val:str, char_to_insert:str) -> bool:
        if char_to_insert.isdigit() and len(current_val) <= 2:
            return True
        return False
    def _validate_alphabetic(self, current_val:str, char_to_insert:str) -> bool:
        if char_to_insert in alphabet:
            return True
        return False
    def _validate_PIN(self, current_val:str, char_to_insert:str) -> bool:
        if char_to_insert.isdigit() and len(current_val) <= 4:
            return True
        return False

if __name__ == "__main__":
    app = InteractionApp()
    app.mainloop()
````
